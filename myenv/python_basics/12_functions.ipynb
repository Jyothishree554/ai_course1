```python
def even_or_odd(num):
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd"
print(even_or_odd(10))  # Output: Even
```

```python
#function with multiple parameters
def add_numbers(a, b):
    return a + b
print(add_numbers(5, 3))  # Output: 8
```

```python
#default parameters
def greet(name="Buddy"):
    return f"Hello, {name}!"
print(greet())  # Output: Hello, Guest!
print(greet("Alice"))  # Output: Hello, Alice!
```

```python
#variable-length arguments
#positional and keyword arguments
#*srgs positional arguments
#**kwargs keyword arguments
def print_info(name, age, *args, **kwargs):
    print(f"Name: {name}, Age: {age}")
    print("Additional Info:", args)
    print("Keyword Info:", kwargs)
print_info("Bob", 30, "Likes Python", "Enjoys hiking", city="New York", profession="Developer")
```

```python
#return statement
def calculate_area(radius):
    import math
    area = math.pi * radius ** 2
    return area
print(calculate_area(5))  # Output: 78.53981633974483

#return multiple values
def min_max(numbers):
    return min(numbers), max(numbers)
nums = [3, 1, 4, 1, 5, 9]
minimum, maximum = min_max(nums)
print(f"Min: {minimum}, Max: {maximum}")  # Output: Min: 1, Max: 9
```
