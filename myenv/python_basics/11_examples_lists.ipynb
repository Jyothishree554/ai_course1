```python
#manage to do list
to_do_list=["do udemy couses", "exercise", "laundry", "cooking"]
print("to_do_list: ",to_do_list)

#adding new task
to_do_list.append("raise PR")
print("after adding new task: ",to_do_list)

#removing completed task
to_do_list.remove("exercise")
print("after removing completed task: ",to_do_list)

if "laundry" in to_do_list:
    print("laundry is still pending")
else:
    print("laundry is completed")

for task in to_do_list:
    print("task: ",task)
```

```python
grades=[85, 90, 78, 92, 88]
total=0
for grade in grades:
    total+=grade
average=total/len(grades)
print("average grade: ",average)
#highest grade
highest=max(grades)
print("highest grade: ",highest)
#lowest grade
lowest=min(grades)
print("lowest grade: ",lowest)
```
