```python
tpl=tuple()
print("empty tuple: ",tpl)
print("type of tpl is: ",type(tpl))
```

```python
#tuples- immutable
my_tuple = (1, 2, 3, 4, 5)
print(my_tuple)
print(type(my_tuple))
#accessing elements
print(my_tuple[0])  # Output: 1
```

```python
#add elements to tuple
my_tuple += (6, 7)
print("updated tuple: ",my_tuple)
```

```python
#type conversion
number=tuple([1,2,3,4,5])
print("tuple created from list: ",number)
print(type(number))

number_list=list([1,2,3,4,50])
print("list created from tuple: ",number_list)
print(type(number_list))
```

```python
#mixed tuple
mixed_tuple=(1,'a',3.14,True)
print("mixed tuple: ",mixed_tuple)
```

```python
#accessing elements in tuple
print("first element of mixed tuple: ",mixed_tuple[0])
print("second element of mixed tuple: ",mixed_tuple[1])
print("last element of mixed tuple: ",mixed_tuple[-1])
```

```python
print(mixed_tuple[0:2])
print(mixed_tuple[::])
print(mixed_tuple[::-1])
```

```python
#tuples operations
tuple1=(1,2,3)
tuple2=(4,5,6)
#concatenation
tuple3=tuple1+tuple2
print("concatenated tuple: ",tuple3)
#repetition
tuple4=tuple1*3
print("repeated tuple: ",tuple4)
#membership test
print(2 in tuple1)  # Output: True
print(5 in tuple1)  # Output: False
#length of tuple
print("length of tuple1: ",len(tuple1))
#count occurrences
print("count of 2 in tuple1: ",tuple1.count(2))
#index of element
print("index of 3 in tuple1: ",tuple1.index(3))
```

```python
#immutabble nature of tuples
try:
    tuple1[0]=10
except TypeError as e:
    print("Error: ",e)
```

```python
#tuple methods
tuple5=(1,2,3,2,4)
print("count of 2 in tuple5: ",tuple5.count(2))
print("index of 3 in tuple5: ",tuple5.index(3))
print("index of 2 in tuple5: ",tuple5.index(2))  # Output: 1 (first occurrence of 2)
```

```python
#packing and unpacking tuples
packed_tuple=1,2,3
print("packed tuple: ",packed_tuple)
a,b,c=packed_tuple
print("unpacked values: ",a,b,c)

#unpacking with *
numbers=(1,2,3,4,5)
a,*b,c,d=numbers
print("first element: ",a)
print("middle elements: ",b)
print("last element: ",c)
```

```python
#nested tuples
nested_tuple=(1,(2,3),4)
print("nested tuple: ",nested_tuple)
print("accessing nested tuple: ",nested_tuple[1][0])  # Output: 2
print("accessing nested tuple: ",nested_tuple[0])  # Output: 1

#nested list
nested_list=[1,[2,3,5,6,7],4]
print("nested list: ",nested_list)
print("accessing nested list: ",nested_list[1][0])  # Output: 2
print("accessing nested list: ",nested_list[0])  # Output: 1
print("accessing nested list: ",nested_list[1][1:3])  # Output: 3
```

```python
#nested tuples
nested_tuple=(1,(2,3),4)
print("nested tuple: ",nested_tuple)
print("accessing nested tuple: ",nested_tuple[1][0])  # Output:
```

```python
#iterating over nested tuple
for item in nested_tuple:
    if isinstance(item, tuple):
        for sub_item in item:
            print("sub item: ",sub_item)
    else:
        print("item: ",item)

for item in nested_tuple:
    print("item: ",item)
```
