```python
#dictionaries
#create a dictionary
empty_dict={}
#or empty_dict=dict()
print("empty_dict: ",empty_dict)
my_dict={'name':'Alice','age':30,'city':'New York'}
print("my_dict: ",my_dict)
print("type of my_dict is: ",type(my_dict))
print("name is ",my_dict['name'])
```

```python
my_dict={'name':'Alice','age':30,'city':'New York', 'name':'Bob'}
print("name is: ",my_dict['name'])
```

```python
#accessing nested dictionary
my_dict={'name':'Alice','age':30,'city':'New York','hobbies':['reading','traveling','cooking']}
print("my_dict with list as value: ",my_dict)
print("hobbies are: ",my_dict['hobbies'][1])

#accessing using get method
print("name using get method: ",my_dict.get('name'))
print("hobbies using get method: ",my_dict.get('hobbies'))
print("non_existent_key using get method: ",my_dict.get('last_name')) #returns None if key does not exist
print("non_existent_key with default value: ",my_dict.get('last_name','Not Found')) #returns default value if key does not exist
```

```python
#modify dictionary
my_dict['age']=31
print("modified age: ",my_dict['age'])
my_dict['country']='USA'
print("added country: ",my_dict['country'])

#add new key-value pair
my_dict['email']='bob@gmail.com'
print("added email: ",my_dict)
```

```python
#delete key-value pair
del my_dict['city']
print("after deleting city: ",my_dict)
```

```python
#dictionaries methods
keys=my_dict.keys()
print("keys: ",keys)
values=my_dict.values() 
print("values: ",values)
items=my_dict.items() #get key-value pairs as tuples
print("items: ",items) #list of tuples

```

```python
#Issue with copy
my_dict_copy=my_dict
print("my_dict_copy: ",my_dict_copy)
print("my_dict", my_dict)

my_dict["name"]="Charlie"
print("after modifying name in my_dict_copy: ")
print("my_dict_copy: ",my_dict_copy)
print("my_dict: ",my_dict)

```

```python
#shallow copy
my_dict_copy=my_dict.copy()
print("my_dict_copy: ",my_dict_copy)
my_dict_copy["name"]="David"
print("after modifying name in my_dict_copy: ")
print("my_dict_copy: ",my_dict_copy)
print("my_dict: ",my_dict)
```

```python
#iterating dictionary
for key in my_dict:
    print("key: ",key)
for key in my_dict.keys():
    print("key: ",key)

for value in my_dict.values():
    print("value: ",value)
for key, value in my_dict.items():
    print("key: ",key, "value: ",value)
```

```python
#nested dictionary
person={
    'name':'Alice',
    'age':30,
    'address':{
        'street':'123 Main St',
        'city':'New York',
        'country':'USA'
    }
}
print("person: ",person)
print("person's city: ",person['address']['city'])
```

```python
students={
    'student1':{
        'name':'Bob',
        'age':20
    },
    'student2':{
        'name':'Charlie',
        'age':22
    }
}
print("students: ",students)
print("student1's name: ",students['student1']['name'])
```

```python
#accessing nested dictionary with get method
print("student1's name using get method: ",students.get('student1',{}).get('name'))
print("student1's name using get method with default value: ",students.get('student1').get('name','Not Found'))

```

```python
#dictionary comprehension
squares={x:x**2 for x in range(1,6)}
print("squares: ",squares)

#conditional dictionary comprehension
even_squares={x:x**2 for x in range(1,11) if x%2==0}
print("even_squares: ",even_squares)
```

```python
#practical example
#use dictionary to count frequency of characters in a string
text="hello world"  
char_freq={}
for char in text:
    if char in char_freq:
        char_freq[char]+=1
    else:
        char_freq[char]=1
print("character frequency: ",char_freq)
```

```python
numbers=[1,2,2,3,3,3,4,4,4,4]
number_freq={}
for num in numbers:
    if num in number_freq:
        number_freq[num]+=1
    else:
        number_freq[num]=1
print("number frequency: ",number_freq)
```

```python
#merge 2 dictionaries
dict1={'a':1,'b':2}
dict2={'b':3,'c':4}
merged_dict={**dict1, **dict2} #values from dict2 will overwrite dict1 for duplicate keys
print("merged_dict: ",merged_dict)

merged_dict2={**dict2, **dict1} #values from dict1 will overwrite dict2 for duplicate keys
print("merged_dict2: ",merged_dict2)
```
