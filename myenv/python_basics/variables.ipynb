```python
#variable types-type of variable is determined at runtime based on the value assigned to it
x=10
print(type(x))

x='Hello'
print(type(x))  

x=2.5
print(type(x))

x=True
print(type(x))
```

```python
#type checking and conversion
type(x)  #check the type of variable x
x=10
print(type(x))

#type conversion
x_str=str(x)  #convert integer to string
print(x_str)
print(type(x_str))

```

```python
age=25
age_str=str(age)  #convert integer to string
print(age_str)
print(type(age_str))

name='Esha'
name_int=int(name)  #this will raise an error because 'Esha' cannot be converted to an integer
```

```python
height=5.6
height_int=int(height)  #this will convert 5.6 to 5 by truncating the decimal part
print(height_int)
print(type(height_int))

print(float(height_int))  #convert integer back to float
```

```python
#dynamic typing
var=10
print(var, type(var))
var='Hello'
print(var, type(var))   
var=3.14
print(var, type(var))
```

```python
#input
name=input("Enter your name: ")
print("Hello, " + name + "!")
print(type(name))  #input() function always returns a string

age=input("Enter your age: ")
print(type(age))  #age will be a string, even if the user enters a number
print("You are " + age + " years old.")

age=int(input("Enter your age: "))  #convert the input string to an integer
print(type(age))
```
