```python
#arithmetic operations
a=20
b=4
add_result=a+b
print("The sum of a and b is:", add_result)
sub_result=a-b
print("The difference of a and b is:", sub_result)      
mul_result=a*b
print("The product of a and b is:", mul_result)
div_result=a/b
print("The quotient of a and b is:", div_result)
mod_result=a%b
print("The modulus of a and b is:", mod_result)
float_div_result=a//b
print("The floor division of a and b is:", float_div_result)
exponent_result=a**b ##floor division is used to get the quotient without the decimal part
print("The exponent of a and b is:", exponent_result)
```

```python
#comparison
#==
str1="Hello"
str2="World"
print("Is str1 equal to str2?", str1 == str2)

str3="Hello"
str4="Hello"
print("Is str1 equal to str2?", str1 == str2)

str5="Hello"
str6="hello"
print("Is str1 equal to str2?", str5 == str6) #case sensitive comparison
```

```python
#!=
print("Is str1 not equal to str2?", str1 != str2)
```

```python
#>
num1=10
num2=5
print("Is num1 greater than num2?", num1 > num2)
```

```python
#<
print("Is num1 less than num2?", num1 < num2)
```

```python
#>=
print("Is num1 greater than or equal to num2?", num1 >= num2)
```

```python
#<=
print("Is num1 less than or equal to num2?", num1 <= num2)
```

```python
# logical operators- and, not, or
x=True
y=False

result=x and y
print("The result of x and y is:", result)

result=x or y
print("The result of x or y is:", result)

result=not x
print("The result of not x is:", result)
```
