```python
lst=[]
print("type of lst is: ",type(lst))

names=['Alice','Bob','Charlie']
print("Names are: ",names)

mixed_list=[1,'Alice',3.14,True]
print("mixed list is: ",mixed_list)

#accessing list elements
fruits=['apple','banana','cherry','date', "mango"]
print("1st element",fruits[0]) #first element
print("2nd element",fruits[1]) #second element
print("last element from list: ",fruits[-1]) #last element

print("slicing list: ",fruits[1:]) #slicing list
print("slicing list: ",fruits[:3]) #slicing list
print("slicing list: ",fruits[1:3]) #slicing list
print("slicing list: ",fruits[-1:-3])
```

```python
#modifying list elements
fruits[0]='avocado'
print("modified list: ",fruits)

fruits[1:]=['blueberry','coconut']
print("modified list: ",fruits)

fruits[1:]=['apple']
print("modified list: ",fruits)

# fruits[1:]="berry"
# print("modified list: ",fruits)

fruits.append('banana')
print("modified list: ",fruits)

fruits.insert(1,'orange')
print("modified list: ",fruits)

fruits.remove('apple')
print("modified list: ",fruits)
```

```python
fruits=['apple','banana','banana','cherry','date', "mango"]
#remove first occurrence of 'banana'
fruits.remove('banana')

print("modified list: ",fruits)

#remove and return last element
last_fruit=fruits.pop()
print("last fruit: ",last_fruit)
print("modified list: ",fruits)
        
```

```python
fruits=['apple','banana','banana','cherry','date', "mango"]
#remove all occurrences of 'banana'
while 'banana' in fruits:
    fruits.remove('banana')
print("modified list: ",fruits)
```

```python
fruits=['apple','banana','cherry','date', "mango"]
fruits.pop(1)
print("modified list: ",fruits)
```

```python
index=fruits.index('cherry')
print("index of cherry: ",index)
```

```python
fruits.insert(1,'banana')
print("modified list: ",fruits)
print(fruits.count('banana'))
```

```python
#sort list
fruits=['mango','apple','cherry','date', "banana"]
fruits.sort()
print("sorted list: ",fruits)
fruits.sort(reverse=True)
print("sorted list in reverse order: ",fruits)
fruits.reverse()
print("reversed list: ",fruits)
print(fruits[0])

fruits.clear()
print("cleared list: ",fruits)
```
