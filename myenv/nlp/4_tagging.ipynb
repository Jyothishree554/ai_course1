```python
#parts of speech (pos) tags:
#n- noun
#v- verb    
#a- adjective
#r- adverb
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
lemmatizer=WordNetLemmatizer()
sentences=nltk.sent_tokenize("I am going to the park. It is a nice day. I will play football with my friends and also enjoy by singing songs.")
for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[lemmatizer.lemmatize(word) for word in words if word not in stopwords.words('english')]
    print(words)
    print(nltk.pos_tag(words))
```

```python
#named entity recognition
sentences="Apple is looking at buying U.K. startup for $1 billion"
words=nltk.word_tokenize(sentences)
print(nltk.pos_tag(words))
import nltk
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
print(nltk.ne_chunk(nltk.pos_tag(words)))
nltk.ne_chunk(nltk.pos_tag(words)).draw()
```
