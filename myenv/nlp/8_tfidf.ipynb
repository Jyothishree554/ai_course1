```python
import pandas as pd
messages=pd.read_csv('messages.csv', sep='\t', names=['label', 'message'])
print(messages)
#Data cleaning and preprocessing
import re
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
wordlemmatize=WordNetLemmatizer()
corpus=[]

for i in range(0, len(messages)):
    review=re.sub('[^a-zA-Z]', ' ', messages['label'][i])  # Fixed: use 'message' not 'label'
    review=review.lower()
    review=review.split()
    review=[wordlemmatize.lemmatize(word) for word in review if not word in stopwords.words('english')]
    review=' '.join(review)
    corpus.append(review)
print(corpus)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer(max_features=100)
X=tfidf.fit_transform(corpus).toarray()

import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000, formatter=dict(float=lambda x: "%.3g" % x))
print("word importance",X)
```
