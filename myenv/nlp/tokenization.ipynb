```python
#tokenization 
#convert paragraph into sentences
import nltk
#nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize
# Tokenize the corpus into sentences using NLTK
corpus="""Hello welcome to the world of programming! Programming is fun and rewarding. 
Let's learn programming together!"""
sentences=sent_tokenize(corpus)
for sentence in sentences:
    print(sentence)
```

```python
#tokenization
#convert paragraph into words
from nltk.tokenize import word_tokenize
# Tokenize the corpus into words using NLTK
corpus="""Hello welcome to the world of programming! Programming is fun and rewarding. 
Let's learn programming together!"""
words=word_tokenize(corpus)
print(words)
print(type(words))
```

```python
from nltk.tokenize import wordpunct_tokenize
# Tokenize the corpus into words using NLTK
corpus="""Hello welcome to the world of programming! Programming is fun and rewarding.
Let's learn programming together!"""
words=wordpunct_tokenize(corpus)
print(words)
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer=TreebankWordTokenizer()
corpus="""Hello welcome to the world of programming! Programming is fun and rewarding.
Let's learn programming together!"""
words=tokenizer.tokenize(corpus)
print(words)
```
