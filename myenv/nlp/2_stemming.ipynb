```python
import nltk
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
```

```python
#stemming
words=["python","pythoner","pythoning","pythoned","pythonly","history","goes"]
#porter stemming
from nltk.stem import PorterStemmer
porter_stemmer=PorterStemmer()
for word in words:
    print(porter_stemmer.stem(word))
```

```python
porter_stemmer.stem("congratualtions")
```

```python
#regexp stemmer
from nltk.stem import RegexpStemmer
reg_stemmer=RegexpStemmer('ing$|s$|able$|e$',min=4)
for word in words:
    print(reg_stemmer.stem(word))
```

```python
reg_stemmer.stem("standing")
```

```python
#snowball stemmer
from nltk.stem import SnowballStemmer
snowball_stemmer=SnowballStemmer('english')
for word in words:
    print(snowball_stemmer.stem(word))

print(snowball_stemmer.stem("fairly"))
print(snowball_stemmer.stem("sportingly"))
print(snowball_stemmer.stem("goes"))
```

```python
#lemmatization- "lemma"
#can be used for chatbot, Q&A, text summarization, sentiment analysis, etc.

from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
lemmatizer.lemmatize("goes",pos='v')
lemmatizer.lemmatize("fairly",pos='a')
```
